```
UPDATE MARKETING.FACEBOOK_ADS
  SET ctr = ROUND(CAST(clicks AS float)/CAST(impressions AS float), 4)
```

```
UPDATE MARKETING.FACEBOOK_ADS
  SET avg_cpc = ROUND(CAST(spend AS float)/CAST(clicks AS float), 2)
```

```
UPDATE MARKETING.FACEBOOK_ADS
  SET cpm =ROUND((CAST(spend AS float)/CAST(impressions AS float))*1000, 4)
```
