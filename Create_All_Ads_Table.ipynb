```
-- Welcome to Snowflake Notebooks!
-- Try out a SQL cell to generate some data.
CREATE OR REPLACE TABLE MARKETING.ALL_ADS AS
SELECT DATE, CAMPAIGN_ID, CAMPAIGN_NAME, AD_SET_ID AS AD_GROUP_ID,      AD_SET_NAME AS AD_GROUP_NAME, IMPRESSIONS, CLICKS, SPEND, CONVERSIONS, CTR, AVG_CPC, CPM, 'Facebook' AS CHANNEL
FROM MARKETING.FACEBOOK_ADS
UNION
SELECT DATE, CAMPAIGN_ID, CAMPAIGN_NAME, AD_GROUP_ID,      AD_GROUP_NAME, IMPRESSIONS, CLICKS, COST, CONVERSIONS, CTR, AVG_CPC, CPM, 'Google' AS CHANNEL
FROM MARKETING.GOOGLE_ADS
UNION
SELECT DATE, CAMPAIGN_ID, CAMPAIGN_NAME, ADGROUP_ID AS AD_GROUP_ID,      ADGROUP_NAME AS AD_GROUP_NAME, IMPRESSIONS, CLICKS, COST, CONVERSIONS, CTR, AVG_CPC, CPM, 'Tiktok' AS CHANNEL
FROM MARKETING.TIKTOK_ADS
```

```
SELECT * FROM MARKETING.ALL_ADS
```

```
SELECT *
FROM MARKETING.ALL_ADS
WHERE DATE IS NULL
   OR CAMPAIGN_ID IS NULL
   OR CAMPAIGN_NAME IS NULL
   OR AD_GROUP_ID IS NULL
   OR AD_GROUP_NAME IS NULL
   OR IMPRESSIONS IS NULL
   OR CLICKS IS NULL
   OR SPEND IS NULL
   OR CONVERSIONS IS NULL
   OR CTR IS NULL
   OR AVG_CPC IS NULL
   OR CPM IS NULL
   OR CHANNEL IS NULL
;
```
